## Объектно-ориентированное программирование (ООП)

```python
class User:

    # атрибуты экземпляра
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # метод экземпляра
    def walk(self, place):
        return "{} идёт куда: {}.".format(self.name, place)

    def speak(self):
        return "{} говорит.".format(self.name)

# создаем экземпляр класса
paul = User("Paul", 10)

# вызываем методы экземпляра
print(paul.walk("на работу"))
print(paul.speak())
```

# Matplotlib

```python
import matplotlib.pyplot as plt
```

```python
def f(x):
    return x ** 2

def g(x):
    return x ** 3

x = np.linspace(-1, 1)

sub1 = plt.subplot(1, 2, 1)
sub1.set_title('f(x)')
sub1.plot(x, f(x))

sub2 = plt.subplot(1, 2, 2)
sub2.set_title('g(x)')
sub2.plot(x, g(x))
plt.show()
```

```python
dist1 = np.random.normal(loc=(-1, -1), size=(100, 2))
dist2 = np.random.normal(loc=(1, 1), size=(100, 2))

plt.title('Two distributions')
plt.scatter(dist1[:, 0], dist1[:, 1])
plt.scatter(dist2[:, 0], dist2[:, 1])
plt.show()
```

Описание признаков:

* PassengerId — id пассажира

* Ticket — номер билета
* Fare — тариф (стоимость билета)
* Cabin — номер кабины
* Embarked — порт, в котором пассажир сел на борт (C - Cherbourg, S - Southampton, Q = Queenstown)
